# EDA: transformed node tables

Quick feel for `sample.parquet` and `taxon.parquet` after transform.

```python
import pandas as pd

df_sample = pd.read_parquet("../data/transformed/nodes/sample.parquet")
df_taxon = pd.read_parquet("../data/transformed/nodes/taxon.parquet")
df_phenotype = pd.read_parquet("../data/transformed/nodes/phenotype.parquet")
```

## Sample nodes

```python
print(df_sample.shape)
df_sample.tail()
```

```python
df_sample.isnull().mean().sort_values(ascending=False) * 100
```

## Taxon nodes

```python
print(df_taxon.shape)
df_taxon.head()
```

```python
df_taxon.isnull().mean().sort_values(ascending=False) * 100
```

```python
df_taxon["rank"].value_counts()
```

```python
df_species = df_taxon[df_taxon["rank"] == "species"]
```

```python
df_species.tail()
```

## Phenotype nodes

```python
print(df_phenotype.shape)
df_phenotype.head()
```

```python
df_phenotype.isnull().mean().sort_values(ascending=False) * 100
```

```python
# dedup check: node table must have exactly one row per distinct id
print("total rows:", len(df_phenotype))
print("distinct id:", df_phenotype["id"].nunique())
print("duplicate id rows:", df_phenotype["id"].duplicated().sum())
```
